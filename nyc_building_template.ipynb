Notebook to prototype injestion of NYC Building Shape Data, isertion into DuckDB, creating spatial index, and running simple GIS query.


```python
import duckdb
import polars as pl
from sodapy import Socrata
import requests
```

```python
# Load API key from file
with open('secreto.txt', 'r') as f:
    api_key = f.read().strip()

# implementation in sodapy

# source domain for NYC Open Data on Socrata
socrata_domain = 'data.cityofnewyork.us'

# initialize client
client = Socrata(
    socrata_domain,
    app_token=api_key,  # Changed from None to api_key
    timeout=100
)

# examine object
print(client)
```

```python
url = 'https://data.cityofnewyork.us/resource/5zhs-2jue.json?$limit=2000'
response = requests.get(url)
df = pl.read_json(response.content, infer_schema_length=1000)

# preview data
print(f"shape of data: {df.shape}")
df.head()
```
